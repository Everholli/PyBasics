# Data Preprocessing: Scaling, Encoding, Normalization 

# Concept: What is Data Preprocessing ?

### Data preprocessing transforms raw data into a clean and usable format.

### It improves model performance and training speed.

##Common tasks:

- Handling missing values

- Scaling numerical features

- Encoding categorical variables

- Normalizing feature ranges

## Feature Scaling

Scaling adjusts values so features contribute equally

Common techniques: StandardScaler (Z-score), MinMaxScaler (0-1 scale)



```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
# Eg Dataset
numeric_data = pd.DataFrame({
    'Salary': [30000, 45000, 60000, 80000, 120000],
    'Age': [25, 32, 47, 51, 62]
})
numeric_data
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
# Standard Scaling (Z-score)
std_scaler = StandardScaler()
scaled_std = std_scaler.fit_transform(numeric_data)
df=pd.DataFrame(scaled_std, columns=numeric_data.columns)
print( df)
```

```python
minmax_scaler = MinMaxScaler()
scaled_minmax = minmax_scaler.fit_transform(numeric_data)
df=pd.DataFrame(scaled_minmax, columns=numeric_data.columns)
print("\nStandard Scaled Data:n", df)
```
