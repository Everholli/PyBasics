```python
# Item details
item1_price, item1_qty = 40, 2
item2_price, item2_qty = 30, 3

# Calculate totals
item1_total = item1_price * item1_qty
item2_total = item2_price * item2_qty

# Subtotal
subtotal = item1_total + item2_total
print(subtotal)
```

```python
# Apply discount
discount = 20
final_total = subtotal - discount

print( "Subtotal.", subtotal)
print("Discount amount:", discount)
print( "Final Total", final_total)
```

```python
purchase_amount = int(input("Enter your amount:"))

if purchase_amount >= 5000:
    discount = 0.2 * purchase_amount
    print(f"Discount: ₹{discount} (20% off)")
elif purchase_amount >= 3000:
    discount = 0.1 * purchase_amount
    print(f"Discount: ₹{discount} (10% off)")
else:
    print("No discount. Shop more!")

final_price = purchase_amount - discount
print(f"Final price: ₹{final_price}")
```

```python
# Sum elements in a 3x3 matrix using nested Loops and user input
total = 0
for i in range(1, 4):
    for j in range(1, 4):
        num = int(input(f"Enter element at position [{i}, {j}]: "))
        total += num

print("Total sum of all matrix elements is:", total)
```
